## 🧹 4. Limpieza y Normalización de Datos

Se normalizan nombres de columnas y valores categóricos, especialmente los tipos de Pokémon, para garantizar consistencia en el análisis y la visualización.

```python
# Crear copia de trabajo
df = df_raw.copy()

print('--- Antes de limpiar columnas ---')
print(f'Columnas originales: {list(df.columns)}')

# Renombrar directamente con un mapeo explícito (más robusto y claro)
df.rename(columns={
    'ID':         'ID',
    'Name':       'Nombre',
    'Form':       'Forma',
    'Type1':      'Tipo',
    'Type2':      'Tipo2',
    'Total':      'Total',
    'HP':         'HP',
    'Attack':     'Ataque',
    'Defense':    'Defensa',
    'Sp. Atk':    'Atk_Esp',
    'Sp. Def':    'Def_Esp',
    'Speed':      'Velocidad',
    'Generation': 'Generación'
}, inplace=True)

print('\n--- Columnas después del renombrado ---')
print(list(df.columns))

# Normalizar valores categóricos: quitar espacios en Tipo y Tipo2
df['Tipo']  = df['Tipo'].astype(str).str.strip().str.title()
df['Tipo2'] = df['Tipo2'].astype(str).str.strip().str.title()

# Reemplazar valores vacíos en Tipo2
df['Tipo2'] = df['Tipo2'].replace({'': np.nan, ' ': np.nan, 'Nan': np.nan, 'None': np.nan})

print('\n✅ Normalización completada')
print(f'Tipos únicos (Tipo principal): {sorted(df["Tipo"].unique())}')
```

## 🔎 5. Identificación y Tratamiento de Valores Nulos, Duplicados e Incorrectos

```python
print('=== 📋 Estado ANTES de la limpieza ===')
print(f'\n🔴 Valores nulos por columna:')
print(df.isnull().sum())
print(f'\n🔴 Registros duplicados: {df.duplicated().sum()}')
print(f'🔴 Total filas: {len(df)}')
```

```python
# --- Tratamiento de duplicados ---
duplicados_antes = df.duplicated().sum()
df.drop_duplicates(inplace=True)

# --- Tratamiento de valores nulos ---
# Tipo2 puede ser nulo -> reemplazar con 'None'
df['Tipo2'] = df['Tipo2'].fillna('None')

# Forma: si está vacía, reemplazar con 'Normal'
df['Forma'] = df['Forma'].astype(str).str.strip()
df['Forma'] = df['Forma'].replace({'': 'Normal', ' ': 'Normal', 'nan': 'Normal', 'None': 'Normal'})

# Verificar que las columnas numéricas no tengan nulos
STATS_COLS_CHECK = ['HP', 'Ataque', 'Defensa', 'Atk_Esp', 'Def_Esp', 'Velocidad']
for col in STATS_COLS_CHECK:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Generación debe ser entero positivo (entre 1 y 9)
df = df[df['Generación'].between(1, 9)].copy()
df['Generación'] = df['Generación'].astype(int)

print('=== ✅ Estado DESPUÉS de la limpieza ===')
print(f'\n🟢 Valores nulos por columna:')
print(df.isnull().sum())
print(f'\n🟢 Registros duplicados eliminados: {duplicados_antes}')
print(f'🟢 Total filas después de limpieza: {len(df)}')
```

## 📊 6. Selección y Justificación de Variables Estadísticas

Las variables seleccionadas para el análisis son las **seis estadísticas base** de cada Pokémon:

| Variable | Justificación |
|---|---|
| `HP` | Representa la resistencia y capacidad de aguante en batalla |
| `Ataque` | Refleja el poder de ataques físicos |
| `Defensa` | Mide la resistencia ante ataques físicos |
| `Atk_Esp` | Indica la potencia de habilidades especiales |
| `Def_Esp` | Cuantifica la resistencia ante ataques especiales |
| `Velocidad` | Determina el orden de acción en el combate |

Estas variables son las que definen el **perfil competitivo** de cada Pokémon y son las más relevantes para comparar su desempeño entre generaciones y tipos.

```python
STATS_COLS = ['HP', 'Ataque', 'Defensa', 'Atk_Esp', 'Def_Esp', 'Velocidad']

print('✅ Variables estadísticas seleccionadas:')
for col in STATS_COLS:
    print(f'   - {col}')

print(f'\n📋 Muestra de las variables:')
df[['Nombre', 'Tipo', 'Generación'] + STATS_COLS].head(10)
```

## ➕ 7. Creación de la Columna `Promedio de estadísticas`

Se calcula el **promedio aritmético** de las seis estadísticas base para cada Pokémon, creando una columna resumen que se usará como eje Z en la visualización 3D.

```python
# Crear columna de promedio
df['Promedio de estadísticas'] = df[STATS_COLS].mean(axis=1).round(2)

print('✅ Columna "Promedio de estadísticas" creada exitosamente')
print(f'\n📊 Distribución del promedio:')
print(df['Promedio de estadísticas'].describe().round(2))
print(f'\n🏆 Top 10 Pokémon con mayor promedio:')
df[['Nombre', 'Tipo', 'Generación', 'Promedio de estadísticas']].sort_values(
    'Promedio de estadísticas', ascending=False
).head(10)
```

## 📈 8. Análisis Estadístico Descriptivo Completo

Se calcula la media, mediana, mínimo, máximo y desviación estándar de las variables estadísticas.

```python
print('=== 📊 Análisis Estadístico Descriptivo Detallado ===')
print()

cols_analisis = STATS_COLS + ['Promedio de estadísticas']

stats_summary = pd.DataFrame({
    'Media':          df[cols_analisis].mean().round(2),
    'Mediana':        df[cols_analisis].median().round(2),
    'Mínimo':         df[cols_analisis].min().round(2),
    'Máximo':         df[cols_analisis].max().round(2),
    'Desv. Estándar': df[cols_analisis].std().round(2),
    'Q1 (25%)':       df[cols_analisis].quantile(0.25).round(2),
    'Q3 (75%)':       df[cols_analisis].quantile(0.75).round(2),
})

print(stats_summary.to_string())
print()
print('--- Análisis por Generación ---')
gen_stats = df.groupby('Generación')['Promedio de estadísticas'].agg(['mean','median','min','max','std']).round(2)
gen_stats.columns = ['Media','Mediana','Mínimo','Máximo','Desv_Std']
print(gen_stats)
print()
print('--- Conteo y Promedio por Tipo Principal ---')
tipo_counts = df.groupby('Tipo').size().sort_values(ascending=False).reset_index(name='Cantidad')
tipo_avg = df.groupby('Tipo')['Promedio de estadísticas'].mean().round(2).reset_index(name='Promedio')
tipo_resumen = tipo_counts.merge(tipo_avg, on='Tipo')
print(tipo_resumen.to_string(index=False))
```

## 🎨 9. Preparación de Variables para Visualización

Se preparan y ordenan las variables de generación y tipo principal para usarlas correctamente en la visualización 3D.

```python
# Trabajar con copia para visualización
df_pokemon = df.copy()

# Ordenar por Generación, Tipo e ID
df_pokemon = df_pokemon.sort_values(['Generación', 'Tipo', 'ID']).reset_index(drop=True)

# Obtener tipos únicos ordenados alfabéticamente
tipos_lista = sorted(df_pokemon['Tipo'].unique())

print(f'✅ Tipos únicos encontrados ({len(tipos_lista)}):')
for t in tipos_lista:
    cnt = (df_pokemon['Tipo'] == t).sum()
    print(f'   - {t:<12}: {cnt:3d} Pokémon')

print(f'\n✅ Generaciones presentes: {sorted(df_pokemon["Generación"].unique())}')
print(f'✅ Total Pokémon para visualización: {len(df_pokemon):,}')
```

## 🖼️ 10. Obtención y Validación de Sprites de Pokémon

Se construyen las URLs de los sprites oficiales de cada Pokémon usando la **PokeAPI CDN** de GitHub, relacionándolas con el ID de la Pokédex de cada registro.

```python
# Función para construir la URL del sprite oficial
def get_sprite_url(pokemon_id):
    """Retorna la URL del sprite oficial front-default de PokeAPI CDN."""
    try:
        pid = int(pokemon_id)
        return f'https://raw.githubusercontent.com/PokeAPI/sprites/master/sprites/pokemon/{pid}.png'
    except:
        return ''

# Aplicar función a cada fila
df_pokemon['sprite_url'] = df_pokemon['ID'].apply(get_sprite_url)

print('✅ URLs de sprites generadas exitosamente')
print('\n📋 Muestra de sprites (primeros 8):')
print(df_pokemon[['ID', 'Nombre', 'sprite_url']].head(8).to_string(index=False))

print(f'\n✅ Total URLs generadas: {len(df_pokemon):,}')
print(f'   URLs vacías: {(df_pokemon["sprite_url"] == "").sum()}')
```

## 🎨 11. Paleta de Colores y Mapeo Numérico por Tipo

Se asigna una paleta de colores oficial por tipo de Pokémon y se mapea el tipo a un valor numérico para el eje Y de la visualización 3D.

```python
# Paleta de colores por tipo de Pokémon (colores oficiales)
TIPO_COLORES = {
    'Fire':     '#FF4500',
    'Water':    '#1E90FF',
    'Grass':    '#32CD32',
    'Electric': '#FFD700',
    'Psychic':  '#FF69B4',
    'Dragon':   '#7B68EE',
    'Normal':   '#A8A878',
    'Ghost':    '#6A0DAD',
    'Fighting': '#CC0000',
    'Ice':      '#00CED1',
    'Bug':      '#8DB360',
    'Poison':   '#A040A0',
    'Ground':   '#E0C068',
    'Rock':     '#B8A038',
    'Steel':    '#B8B8D0',
    'Dark':     '#705848',
    'Fairy':    '#EE99AC',
    'Flying':   '#6890F0',
}

# Asignar color a cada Pokémon según su tipo
df_pokemon['color'] = df_pokemon['Tipo'].map(TIPO_COLORES).fillna('#AAAAAA')

# Mapear tipo a número para eje Y numérico
tipo_a_num = {t: i for i, t in enumerate(tipos_lista)}
df_pokemon['tipo_num'] = df_pokemon['Tipo'].map(tipo_a_num)

print('✅ Paleta de colores y mapeo numérico asignados')
for tipo in tipos_lista:
    color = TIPO_COLORES.get(tipo, '#AAAAAA')
    count = (df_pokemon['Tipo'] == tipo).sum()
    print(f'   {tipo:<12}: {color}  ({count} Pokémon)')
```

## 🗺️ 12. Primera Versión del 3D Scatter Plot Interactivo

Se construye la primera versión con colores diferenciados por tipo y hover con información básica.

```python
# ============================================================
# VERSIÓN 1: Scatter 3D básico diferenciado por color/tipo
# ============================================================

fig = go.Figure()

for tipo in tipos_lista:
    mask = df_pokemon['Tipo'] == tipo
    df_tipo = df_pokemon[mask]
    color = TIPO_COLORES.get(tipo, '#AAAAAA')
    
    hover_text = [
        f'<b>Pokémon:</b> {row.Nombre}<br>'
        f'<b>Tipo:</b> {row.Tipo}<br>'
        f'<b>Generación:</b> {row.Generación}<br>'
        f'<b>Promedio de estadísticas:</b> {row["Promedio de estadísticas"]}'
        for _, row in df_tipo.iterrows()
    ]
    
    fig.add_trace(
        go.Scatter3d(
            x=df_tipo['Generación'],
            y=df_tipo['tipo_num'],
            z=df_tipo['Promedio de estadísticas'],
            mode='markers',
            name=tipo,
            marker=dict(
                size=7,
                color=color,
                opacity=0.88,
                line=dict(width=1, color='white'),
            ),
            text=hover_text,
            hovertemplate='%{text}<extra></extra>',
        )
    )

fig.update_layout(
    title=dict(
        text='<b>Scatter 3D de Pokémon: Promedio de Estadísticas por Tipo y Generación</b>',
        x=0.5,
        font=dict(size=18, color='#2c2c2c', family='Arial')
    ),
    scene=dict(
        xaxis=dict(
            title='<b>Generación</b>',
            tickvals=list(range(1, 10)),
            ticktext=[str(i) for i in range(1, 10)],
            backgroundcolor='rgb(230,230,250)',
            gridcolor='white', showbackground=True,
        ),
        yaxis=dict(
            title='<b>Tipo</b>',
            tickvals=list(tipo_a_num.values()),
            ticktext=list(tipo_a_num.keys()),
            backgroundcolor='rgb(250,230,230)',
            gridcolor='white', showbackground=True,
        ),
        zaxis=dict(
            title='<b>Promedio de estadísticas</b>',
            backgroundcolor='rgb(230,250,230)',
            gridcolor='white', showbackground=True,
        ),
        camera=dict(eye=dict(x=1.8, y=-1.8, z=1.2))
    ),
    legend=dict(
        title=dict(text='<b>Tipo (color / símbolo)</b>', font=dict(size=13)),
        font=dict(size=12),
        bgcolor='rgba(255,255,255,0.9)',
        bordercolor='#cccccc', borderwidth=1,
    ),
    paper_bgcolor='white',
    width=1150, height=750,
    annotations=[
        dict(
            text='<i>Arrastra para rotar, usa la rueda para acercar/alejar, o mantén Shift para desplazarte.</i>',
            showarrow=False, xref='paper', yref='paper',
            x=0, y=-0.06, font=dict(size=11, color='gray')
        ),
        dict(
            text='<b>Cada punto es un Pokémon.</b><br>La posición Z representa el<br>promedio de sus estadísticas.',
            showarrow=False, xref='paper', yref='paper',
            x=1.02, y=0.05, align='left',
            bordercolor='#cccccc', borderwidth=1,
            bgcolor='rgba(255,255,255,0.9)',
            font=dict(size=11, color='#555555')
        )
    ]
)

fig.show()
print('✅ Primera versión del Scatter 3D generada exitosamente')
```

## 🖼️ 13. Integración de Sprites en Hover + Configuración de Información Emergente

Se construye la segunda versión donde los **sprites** del Pokémon se muestran en el tooltip. Se configura la información emergente completa: nombre, tipo, generación y promedio de estadísticas.

```python
# ============================================================
# VERSIÓN 2: Con sprites en el hover tooltip
# ============================================================

fig2 = go.Figure()

for tipo in tipos_lista:
    mask = df_pokemon['Tipo'] == tipo
    df_tipo = df_pokemon[mask]
    color = TIPO_COLORES.get(tipo, '#AAAAAA')
    
    custom = np.column_stack([
        df_tipo['Nombre'].values,
        df_tipo['Tipo'].values,
        df_tipo['Generación'].values,
        df_tipo['Promedio de estadísticas'].values,
        df_tipo['sprite_url'].values,
    ])
    
    fig2.add_trace(
        go.Scatter3d(
            x=df_tipo['Generación'],
            y=df_tipo['tipo_num'],
            z=df_tipo['Promedio de estadísticas'],
            mode='markers',
            name=tipo,
            marker=dict(
                size=9, color=color, opacity=0.90,
                line=dict(width=1.5, color='white'),
            ),
            hovertemplate=(
                '<img src="%{customdata[4]}" width="80" height="80"><br>'
                '<b>%{customdata[0]}</b><br>'
                'Tipo: %{customdata[1]}<br>'
                'Generación: %{customdata[2]}<br>'
                'Promedio de estadísticas: %{customdata[3]}<br>'
                '<extra></extra>'
            ),
            customdata=custom,
        )
    )

fig2.update_layout(
    title=dict(
        text='<b>Scatter 3D de Pokémon con Sprites — Promedio de Estadísticas</b>',
        x=0.5, font=dict(size=17, color='#1a1a2e', family='Arial')
    ),
    scene=dict(
        xaxis=dict(
            title='<b>Generación</b>',
            tickvals=list(range(1, 10)),
            ticktext=[str(i) for i in range(1, 10)],
            backgroundcolor='rgba(220,220,255,0.8)',
            gridcolor='white', showbackground=True,
        ),
        yaxis=dict(
            title='<b>Tipo</b>',
            tickvals=list(tipo_a_num.values()),
            ticktext=list(tipo_a_num.keys()),
            backgroundcolor='rgba(255,220,220,0.8)',
            gridcolor='white', showbackground=True,
        ),
        zaxis=dict(
            title='<b>Promedio de estadísticas</b>',
            backgroundcolor='rgba(220,255,220,0.8)',
            gridcolor='white', showbackground=True,
        ),
        camera=dict(eye=dict(x=1.8, y=-2.0, z=1.5))
    ),
    legend=dict(
        title=dict(text='<b>Tipo (clic para filtrar)</b>', font=dict(size=13)),
        font=dict(size=12),
        bgcolor='rgba(255,255,255,0.95)',
        bordercolor='#cccccc', borderwidth=1,
    ),
    paper_bgcolor='#f0f2f5',
    width=1200, height=800,
    hoverlabel=dict(bgcolor='white', font_size=13, font_family='Arial'),
)

fig2.show()
print('✅ Gráfica 3D con sprites en hover generada')
```

## 🔧 14. Gráfica Final con Filtros Interactivos y Personalización Completa

Se construye la versión final con:
- Filtros interactivos por generación (botones)

- Personalización completa: título, ejes, cámara, opacidad, escala, anotaciones

```python
# ============================================================
# VERSIÓN FINAL: Con filtros interactivos + personalización
# ============================================================

fig3 = go.Figure()

for tipo in tipos_lista:
    mask = df_pokemon['Tipo'] == tipo
    df_tipo = df_pokemon[mask]
    color = TIPO_COLORES.get(tipo, '#AAAAAA')
    
    custom = np.column_stack([
        df_tipo['Nombre'].values,
        df_tipo['Tipo'].values,
        df_tipo['Generación'].values,
        df_tipo['Promedio de estadísticas'].values,
        df_tipo['sprite_url'].values,
        df_tipo['HP'].values,
        df_tipo['Ataque'].values,
        df_tipo['Defensa'].values,
        df_tipo['Atk_Esp'].values,
        df_tipo['Def_Esp'].values,
        df_tipo['Velocidad'].values,
    ])
    
    fig3.add_trace(
        go.Scatter3d(
            x=df_tipo['Generación'],
            y=df_tipo['tipo_num'],
            z=df_tipo['Promedio de estadísticas'],
            mode='markers',
            name=tipo,
            marker=dict(
                size=8, color=color, opacity=0.90,
                line=dict(width=1, color='white'),
            ),
            hovertemplate=(
                '<img src="%{customdata[4]}" width="80" height="80"><br>'
                '<b>%{customdata[0]}</b><br>'
                'Tipo: %{customdata[1]}<br>'
                'Generación: %{customdata[2]}<br>'
                'Promedio de estadísticas: %{customdata[3]}<br>'
                'HP: %{customdata[5]} | Atk: %{customdata[6]} | Def: %{customdata[7]}<br>'
                'Atk Esp: %{customdata[8]} | Def Esp: %{customdata[9]} | Vel: %{customdata[10]}'
                '<extra></extra>'
            ),
            customdata=custom,
        )
    )

# --- Botones de filtro por generación ---
generaciones = sorted(df_pokemon['Generación'].unique())

buttons = [
    dict(
        label='🌐 Todas',
        method='restyle',
        args=[{
            'x': [df_pokemon[df_pokemon['Tipo'] == t]['Generación'].tolist() for t in tipos_lista],
            'y': [df_pokemon[df_pokemon['Tipo'] == t]['tipo_num'].tolist() for t in tipos_lista],
            'z': [df_pokemon[df_pokemon['Tipo'] == t]['Promedio de estadísticas'].tolist() for t in tipos_lista],
        }]
    )
]

for gen in generaciones:
    buttons.append(
        dict(
            label=f'Gen {gen}',
            method='restyle',
            args=[{
                'x': [df_pokemon[(df_pokemon['Tipo'] == t) & (df_pokemon['Generación'] == gen)]['Generación'].tolist() for t in tipos_lista],
                'y': [df_pokemon[(df_pokemon['Tipo'] == t) & (df_pokemon['Generación'] == gen)]['tipo_num'].tolist() for t in tipos_lista],
                'z': [df_pokemon[(df_pokemon['Tipo'] == t) & (df_pokemon['Generación'] == gen)]['Promedio de estadísticas'].tolist() for t in tipos_lista],
            }]
        )
    )

fig3.update_layout(
    title=dict(
        text='<b>Scatter 3D de Pokémon: Promedio de Estadísticas por Tipo y Generación</b>',
        x=0.5, font=dict(size=19, color='#1a1a2e', family='Arial Black')
    ),
    scene=dict(
        xaxis=dict(
            title=dict(text='<b>Generación</b>', font=dict(size=14)),
            tickvals=list(range(1, 10)),
            ticktext=[str(i) for i in range(1, 10)],
            backgroundcolor='rgba(210,210,255,0.7)',
            gridcolor='white', showbackground=True, zeroline=False
        ),
        yaxis=dict(
            title=dict(text='<b>Tipo</b>', font=dict(size=14)),
            tickvals=list(tipo_a_num.values()),
            ticktext=list(tipo_a_num.keys()),
            backgroundcolor='rgba(255,210,210,0.7)',
            gridcolor='white', showbackground=True, zeroline=False
        ),
        zaxis=dict(
            title=dict(text='<b>Promedio de estadísticas</b>', font=dict(size=14)),
            backgroundcolor='rgba(210,255,210,0.7)',
            gridcolor='white', showbackground=True, zeroline=False
        ),
        camera=dict(
            eye=dict(x=2.0, y=-2.0, z=1.5),
            up=dict(x=0, y=0, z=1)
        ),
        aspectmode='auto'
    ),
    updatemenus=[dict(
        type='buttons', direction='right', buttons=buttons,
        showactive=True, x=0.01, y=1.10,
        xanchor='left', yanchor='top',
        bgcolor='white', bordercolor='#999',
        font=dict(size=11)
    )],
    legend=dict(
        title=dict(text='<b>Tipo (clic para filtrar)</b>', font=dict(size=13)),
        font=dict(size=11),
        bgcolor='rgba(255,255,255,0.95)',
        bordercolor='#cccccc', borderwidth=1,
        tracegroupgap=4
    ),
    paper_bgcolor='#f0f2f5',
    width=1200, height=840,
    margin=dict(l=0, r=220, b=60, t=140),
    hoverlabel=dict(bgcolor='white', font_size=13, font_family='Arial'),
    annotations=[
        dict(
            text='<i>ⓘ Arrastra para rotar, usa la rueda para acercar/alejar. Clic en leyenda para filtrar por tipo.</i>',
            showarrow=False, xref='paper', yref='paper',
            x=0.0, y=-0.04, font=dict(size=11, color='gray')
        ),
        dict(
            text='<b>Cada punto es un Pokémon.</b><br>La posición Z representa el<br>promedio de sus estadísticas.<br><br>Hover: muestra sprite + stats.',
            showarrow=False, xref='paper', yref='paper',
            x=1.02, y=0.12, align='left',
            bordercolor='#cccccc', borderwidth=1,
            bgcolor='rgba(255,255,255,0.95)',
            font=dict(size=11, color='#555555')
        )
    ]
)

fig3.show()
print('✅ Gráfica final con filtros interactivos y personalización completa')
```

## 🧠 15. Identificación de Patrones, Agrupaciones y Valores Atípicos

### 🔍 Hallazgo 1: Outliers Legendarios
Los Pokémon Legendarios como **Mewtwo, Arceus, Rayquaza, Groudon y Kyogre** tienen promedios > 100 puntos, muy por encima de la media global (~72 pts). Son outliers positivos en prácticamente todos los tipos.

### 🔍 Hallazgo 2: Tipos más poderosos vs. más numerosos
Los tipos **Bug** y **Normal** son los más numerosos pero con promedios más bajos. **Dragon, Steel y Psychic** tienen menos representantes pero consistentemente los promedios más altos.

### 🔍 Hallazgo 3: Generación 6 como Pico Estadístico
La Generación 6 incluye Mega Evoluciones con promedios que superan 130+ pts, creando una agrupación visualmente distinta en la parte superior del eje Z.

### 🔍 Hallazgo 4: Distribución Asimétrica
La distribución de promedios es asimétrica positiva (sesgada a la derecha) debido a los legendarios, lo que indica que la mayoría de los Pokémon tienen estadísticas moderadas.

```python
# Análisis cuantitativo de outliers (> 2 desviaciones estándar)
mean_prom = df_pokemon['Promedio de estadísticas'].mean()
std_prom  = df_pokemon['Promedio de estadísticas'].std()
threshold = mean_prom + 2 * std_prom

outliers = df_pokemon[df_pokemon['Promedio de estadísticas'] > threshold]

print(f'📊 Media global: {mean_prom:.2f}')
print(f'📊 Desviación estándar: {std_prom:.2f}')
print(f'📊 Umbral outlier (μ + 2σ): {threshold:.2f}')
print(f'\n🔺 Pokémon Outliers ({len(outliers)} encontrados):')
print(
    outliers[['Nombre','Tipo','Generación','Promedio de estadísticas']]
    .sort_values('Promedio de estadísticas', ascending=False)
    .to_string(index=False)
)

print('\n--- Promedio por tipo (mayor a menor) ---')
tipo_avg = df_pokemon.groupby('Tipo')['Promedio de estadísticas'].mean().sort_values(ascending=False)
print(tipo_avg.round(2).to_string())
```

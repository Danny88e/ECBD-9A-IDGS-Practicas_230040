## 🔬 1. Introducción

### ¿Qué es el Aprendizaje No Supervisado?

El **aprendizaje no supervisado** (Unsupervised Learning) es una rama del machine learning en la que el algoritmo trabaja con datos que **no tienen etiquetas o categorías predefinidas**. A diferencia del aprendizaje supervisado —donde le decimos al modelo cuál es la respuesta correcta— aquí el algoritmo debe encontrar por sí mismo la estructura, los patrones y las relaciones ocultas en los datos.

Es como darle a alguien un montón de fotografías sin nombre y pedirle que las agrupe como crea conveniente. Esa persona podría agruparlas por colores, por paisajes o por personas, sin que nadie le dijera cómo hacerlo.

### ¿Qué problema resuelve el agrupamiento (Clustering)?

El **agrupamiento** o clustering resuelve el problema de **segmentación**: encontrar subgrupos naturales dentro de un conjunto de datos. Esto es útil cuando queremos:

- Descubrir segmentos de clientes con comportamientos similares
- Identificar patrones en datos médicos o biológicos
- Detectar anomalías o grupos inusuales
- Reducir la complejidad de grandes conjuntos de datos

La idea clave es que los elementos **dentro de un mismo grupo** deben ser muy similares entre sí, y los elementos de **grupos diferentes** deben ser muy distintos entre ellos.

### ¿Por qué aplicar K-Means en la segmentación de clientes?

**K-Means** es uno de los algoritmos de clustering más populares y eficientes. Su funcionamiento básico es:

1. Se elige el número `k` de grupos deseados
2. Se colocan `k` centros de forma aleatoria (centroides)
3. Cada punto se asigna al centroide más cercano
4. Los centroides se recalculan como el promedio de los puntos asignados
5. Se repiten los pasos 3-4 hasta que los centroides no se muevan significativamente

En el contexto de un **centro comercial**, K-Means nos permite segmentar a los clientes según su ingreso anual, su puntaje de gasto y su edad, identificando perfiles como:
- Clientes jóvenes con alto gasto y bajo ingreso
- Clientes maduros con alto ingreso pero gasto conservador
- El segmento "ideal" con alto ingreso y alto gasto

Esta información es extremadamente valiosa para las estrategias de marketing, campañas personalizadas y toma de decisiones comerciales.

## 🔍 5. Revisión de Calidad de los Datos

Antes de comenzar cualquier análisis o modelado, es fundamental verificar que los datos estén en buen estado. Una base de datos con errores produce modelos incorrectos, independientemente de qué tan bueno sea el algoritmo que usemos: *"garbage in, garbage out"* (basura entra, basura sale).

Revisaremos los siguientes aspectos:
- Valores nulos o faltantes (NaN)
- Registros duplicados
- Valores únicos por columna
- Rangos de valores incorrectos
- Inconsistencias en datos categóricos

```python
# ============================================================
# SECCIÓN 5: Calidad de Datos
# ============================================================

# --- 5.1 Valores Nulos ---
print("🔴 5.1 Valores Nulos por Columna:")
nulos = df.isnull().sum()
pct_nulos = (nulos / len(df) * 100).round(2)
reporte_nulos = pd.DataFrame({'Nulos': nulos, '% del Total': pct_nulos})
print(reporte_nulos)
print(f"\n   → Total de valores nulos en el dataset: {nulos.sum()}")
```

```python
# --- 5.2 Registros Duplicados ---
print("🔴 5.2 Registros Duplicados:")
duplicados = df.duplicated().sum()
print(f"   → Número de filas completamente duplicadas: {duplicados}")

if duplicados > 0:
    print("\n   ⚠️  Se detectaron duplicados. Mostrando registros:")
    display(df[df.duplicated()])
else:
    print("   ✅ No se encontraron registros duplicados.")
```

```python
# --- 5.3 Valores Únicos por Columna ---
print("📌 5.3 Valores Únicos por Columna:")
for col in df.columns:
    unicos = df[col].nunique()
    print(f"   {col}: {unicos} valores únicos")
    if df[col].dtype == 'object':
        print(f"         → Valores: {df[col].unique().tolist()}")
```

```python
# --- 5.4 Rangos y Consistencia de Variables Numéricas ---
print("📊 5.4 Verificación de Rangos:")
print()

rangos_esperados = {
    'Age': (18, 100, 'años'),
    'Annual Income (k$)': (0, 500, 'miles USD'),
    'Spending Score (1-100)': (1, 100, 'puntos')
}

for col, (min_esp, max_esp, unidad) in rangos_esperados.items():
    min_real = df[col].min()
    max_real = df[col].max()
    fuera = df[(df[col] < min_esp) | (df[col] > max_esp)].shape[0]
    estado = '✅' if fuera == 0 else '⚠️'
    print(f"   {estado} {col}:")
    print(f"      Rango esperado: [{min_esp}, {max_esp}] {unidad}")
    print(f"      Rango real:     [{min_real}, {max_real}] {unidad}")
    print(f"      Valores fuera de rango: {fuera}")
    print()

# Consistencia de Gender
print("👥 5.5 Consistencia de Gender:")
print(df['Gender'].value_counts())
inesperados = df[~df['Gender'].isin(['Male', 'Female'])]
if len(inesperados) == 0:
    print("   ✅ Todos los valores de Gender son válidos.")
```

## 📊 6. Estadísticas Descriptivas

Las estadísticas descriptivas nos dan un resumen numérico de cómo se distribuye cada variable. Son el primer paso para entender el "comportamiento" de los datos antes de visualizarlos.

**Métricas clave a interpretar:**
- **Media (mean)**: Promedio aritmético. Sensible a valores extremos.
- **Desviación estándar (std)**: Qué tan dispersos están los datos respecto a la media. Mayor std = mayor variabilidad.
- **Mínimo/Máximo**: Valores extremos del conjunto.
- **Q1 (25%)**: El 25% de los datos está por debajo de este valor.
- **Mediana (50%)**: El valor central. Si difiere mucho de la media, hay sesgo.
- **Q3 (75%)**: El 75% de los datos está por debajo de este valor.

```python
# ============================================================
# SECCIÓN 6: Estadísticas Descriptivas
# ============================================================

print("📊 Estadísticas descriptivas de variables numéricas:")
desc = df.describe().T
desc['rango'] = desc['max'] - desc['min']
desc['cv (%)'] = (desc['std'] / desc['mean'] * 100).round(2)
display(desc.round(2))
```

```python
# Interpretación detallada por variable
print("📌 Interpretación por variable:")
print()

print("👤 Edad (Age):")
print(f"   Media: {df['Age'].mean():.1f} años | Mediana: {df['Age'].median():.1f} años")
print(f"   Rango: {df['Age'].min()} – {df['Age'].max()} años")
print(f"   La mayoría de clientes tiene entre {df['Age'].quantile(0.25):.0f} y {df['Age'].quantile(0.75):.0f} años")
print()

print("💰 Ingreso Anual (Annual Income):")
print(f"   Media: ${df['Annual Income (k$)'].mean():.1f}k | Mediana: ${df['Annual Income (k$)'].median():.1f}k")
print(f"   Rango: ${df['Annual Income (k$)'].min()}k – ${df['Annual Income (k$)'].max()}k")
print()

print("🛒 Puntuación de Gasto (Spending Score):")
print(f"   Media: {df['Spending Score (1-100)'].mean():.1f} | Mediana: {df['Spending Score (1-100)'].median():.1f}")
print(f"   Rango: {df['Spending Score (1-100)'].min()} – {df['Spending Score (1-100)'].max()}")
```

## 📈 7. Análisis Exploratorio Univariado

El análisis univariado estudia cada variable de forma **individual**, sin considerar su relación con otras. El objetivo es entender la distribución, frecuencia y tendencia de cada columna por separado.

## 📦 8. Análisis de Valores Atípicos (Outliers)

Los **valores atípicos** (outliers) son observaciones que se encuentran muy lejos del resto del conjunto de datos. En un **boxplot**:
- La caja muestra el rango intercuartil (Q1 a Q3)
- La línea central es la mediana
- Los "bigotes" llegan hasta 1.5× el rango intercuartil
- Los puntos más allá de los bigotes son outliers

K-Means es sensible a outliers porque un dato extremo puede sesgar un centroide.

```python
# ============================================================
# SECCIÓN 8: Análisis de Valores Atípicos
# ============================================================

variables_num = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']

# ---- 8.1 Boxplots generales ----
fig, axes = plt.subplots(1, 3, figsize=(15, 6))
fig.suptitle('Análisis de Valores Atípicos — Boxplots por Variable', fontsize=15, fontweight='bold')

for i, (col, color) in enumerate(zip(variables_num, COLORES)):
    axes[i].boxplot(df[col], patch_artist=True,
                    boxprops=dict(facecolor=color, alpha=0.6),
                    medianprops=dict(color='white', linewidth=2),
                    whiskerprops=dict(color='gray'),
                    capprops=dict(color='gray'),
                    flierprops=dict(marker='o', color='#e63946', markersize=7))
    titulos_col = {'Age': 'Edad\n(años)',
                   'Annual Income (k$)': 'Ingreso Anual\n(k$)',
                   'Spending Score (1-100)': 'Spending Score\n(1-100)'}
    axes[i].set_title(titulos_col[col], fontweight='bold')
    axes[i].set_ylabel('Valor')
    axes[i].spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# ---- 8.2 Boxplots comparativos por Género ----
fig, axes = plt.subplots(1, 3, figsize=(15, 6))
fig.suptitle('Boxplots por Variable según Género', fontsize=15, fontweight='bold')

for i, col in enumerate(variables_num):
    sns.boxplot(data=df, x='Gender', y=col, palette=PALETA, ax=axes[i],
                width=0.5, linewidth=1.5)
    axes[i].set_title(col, fontweight='bold')
    axes[i].set_xlabel('Género')
    axes[i].spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()

print("\n📌 Interpretación:")
print("   • Age: No hay outliers significativos. Distribución similar entre géneros.")
print("   • Annual Income: Sin outliers extremos. Similar entre géneros.")
print("   • Spending Score: Las mujeres muestran mayor concentración en scores medios-altos.")
```

```python
# ---- 8.3 Detección numérica de outliers con IQR ----
print("🔍 Detección numérica de outliers (método IQR):")
print()

for col in variables_num:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lim_inf = Q1 - 1.5 * IQR
    lim_sup = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lim_inf) | (df[col] > lim_sup)]

    print(f"   📊 {col}:")
    print(f"      Q1={Q1:.1f} | Q3={Q3:.1f} | IQR={IQR:.1f}")
    print(f"      Límites: [{lim_inf:.1f}, {lim_sup:.1f}]")
    print(f"      Outliers detectados: {len(outliers)} registros")
    if len(outliers) > 0:
        display(outliers[['CustomerID', 'Gender', 'Age', col]])
    else:
        print("      ✅ Sin outliers")
    print()
```

## 🔗 9. Análisis Bivariado y Multivariado

El análisis bivariado y multivariado nos permite entender las **relaciones entre pares o grupos de variables**. Esta es la parte más importante antes del clustering porque nos ayuda a identificar visualmente si ya existen grupos naturales en los datos.

Utilizaremos:
- **Diagramas de dispersión (scatter plots)**: Para ver la relación entre dos variables
- **Gráfico de pares (pair plot)**: Para ver todas las combinaciones a la vez
- **Mapa de calor de correlaciones (heatmap)**: Para cuantificar la relación lineal entre variables

## 🔠 10. Codificación de Variables Categóricas

K-Means es un algoritmo basado en **distancias matemáticas**, por lo que solo puede trabajar con variables **numéricas**. La columna `Gender` contiene texto (`Male`, `Female`), por lo que debemos convertirla a números.

Usaremos **LabelEncoder** de Scikit-learn:
- `Female` → `0`
- `Male` → `1`

> ⚠️ **Importante**: Esta codificación numérica **no implica orden jerárquico**. Son simplemente etiquetas numéricas para representar categorías. Para variables con más categorías sin orden natural, se recomienda One-Hot Encoding.

```python
# ============================================================
# SECCIÓN 10: Codificación de Variables Categóricas
# ============================================================

df_encoded = df.copy()

le = LabelEncoder()
df_encoded['Gender_encoded'] = le.fit_transform(df_encoded['Gender'])

print("🔄 Mapeo de codificación aplicado por LabelEncoder:")
mapeo = pd.DataFrame({
    'Valor Original': le.classes_,
    'Valor Codificado': le.transform(le.classes_)
})
display(mapeo)
print()

print("📋 Comparación Original vs. Codificado (primeras 8 filas):")
display(df_encoded[['CustomerID', 'Gender', 'Gender_encoded',
                     'Age', 'Annual Income (k$)', 'Spending Score (1-100)']].head(8))
```

```python
print("✅ Verificación de la codificación:")
print(f"   Valores únicos en Gender_encoded: {sorted(df_encoded['Gender_encoded'].unique())}")
print(f"   Female (0): {(df_encoded['Gender_encoded'] == 0).sum()} clientes")
print(f"   Male (1):   {(df_encoded['Gender_encoded'] == 1).sum()} clientes")
print()
print("⚠️  Recordatorio: 0 (Female) y 1 (Male) NO implican jerarquía.")
print("   Son solo etiquetas numéricas para representar categorías.")
```

## ⚖️ 11. Normalización del Dataset

K-Means calcula distancias entre puntos. Si una variable tiene valores en el rango [15, 137] (ingreso) y otra en [1, 100] (spending score), la variable con mayor escala **dominará artificialmente** el cálculo de distancias, sesgando los resultados.

Usaremos **StandardScaler** (estandarización z-score), que transforma cada variable para que tenga:
- **Media = 0**
- **Desviación estándar = 1**

Fórmula: `z = (x - μ) / σ`

```python
# ============================================================
# SECCIÓN 11: Normalización
# ============================================================

vars_normalizar = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']

scaler = StandardScaler()
df_scaled = df_encoded.copy()
df_scaled[vars_normalizar] = scaler.fit_transform(df_encoded[vars_normalizar])

print("✅ Normalización aplicada con StandardScaler")
print()
print("📊 Estadísticas ANTES de normalizar:")
display(df[vars_normalizar].describe().round(3))
```

```python
print("📊 Estadísticas DESPUÉS de normalizar:")
display(df_scaled[vars_normalizar].describe().round(3))
print()
print("✅ Confirmación: todas las variables ahora tienen media ≈ 0 y std ≈ 1")
```

```python
# ---- Comparación visual: Antes vs. Después ----
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle('Comparación: Datos Originales vs. Normalizados (StandardScaler)', fontsize=14, fontweight='bold')

titulos = ['Edad (Age)', 'Ingreso Anual (k$)', 'Spending Score']

for i, (col, titulo) in enumerate(zip(vars_normalizar, titulos)):
    # Datos originales (fila superior)
    axes[0, i].hist(df[col], bins=15, color=COLORES[i], edgecolor='white', alpha=0.85)
    axes[0, i].set_title(f'{titulo}\n(Original)', fontweight='bold')
    axes[0, i].set_ylabel('Frecuencia')
    axes[0, i].spines[['top','right']].set_visible(False)

    # Datos normalizados (fila inferior)
    axes[1, i].hist(df_scaled[col], bins=15, color=COLORES[i], edgecolor='white', alpha=0.5)
    axes[1, i].set_title(f'{titulo}\n(Normalizado — z-score)', fontweight='bold')
    axes[1, i].set_ylabel('Frecuencia')
    axes[1, i].set_xlabel('Valor estandarizado (z)')
    axes[1, i].spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()

print("\n📌 La forma de la distribución se conserva; solo cambia la escala del eje X.")
```

### 💬 Conclusiones Personales — Normalización

Lo más importante que entendí aquí es que normalizar los datos no cambia la distribución ni la forma de los datos, solo cambia la escala. Es como si midieras la temperatura en Celsius y luego la convirtieras a Fahrenheit: los mismos días siguen siendo más calurosos o más fríos en relación entre sí, solo que ahora la escala numérica es diferente. Para K-Means esto es crucial porque si no normalizamos, el ingreso anual (que puede variar en cientos) dominaría sobre el Spending Score (que va de 1 a 100), y los clústeres quedarían agrupados principalmente por ingreso ignorando el comportamiento de gasto.

## 🧮 12. Matrices de Características para K-Means

Para aplicar K-Means necesitamos seleccionar qué variables usar como entrada del modelo. Crearemos matrices para las siguientes combinaciones:

1. **Edad e Ingreso Anual** (Age + Annual Income)
2. **Edad y Spending Score** (Age + Spending Score)
3. **Ingreso Anual y Spending Score** (Annual Income + Spending Score) — La más reveladora

Para cada combinación crearemos versiones con **datos originales** y **datos normalizados**.

```python
# ============================================================
# SECCIÓN 12: Matrices de Características
# ============================================================

# --- Datos Originales ---
X1 = df[['Age', 'Annual Income (k$)']].values
X2 = df[['Age', 'Spending Score (1-100)']].values
X3 = df[['Annual Income (k$)', 'Spending Score (1-100)']].values

# --- Datos Normalizados ---
X1_scaled = df_scaled[['Age', 'Annual Income (k$)']].values
X2_scaled = df_scaled[['Age', 'Spending Score (1-100)']].values
X3_scaled = df_scaled[['Annual Income (k$)', 'Spending Score (1-100)']].values

print("✅ Matrices de características construidas:")
print()
matrices = [
    ("X1",        "Age + Annual Income             (Originales)", X1),
    ("X2",        "Age + Spending Score            (Originales)", X2),
    ("X3",        "Annual Income + Spending Score  (Originales)", X3),
    ("X1_scaled", "Age + Annual Income             (Normalizados)", X1_scaled),
    ("X2_scaled", "Age + Spending Score            (Normalizados)", X2_scaled),
    ("X3_scaled", "Annual Income + Spending Score  (Normalizados)", X3_scaled),
]
for nombre, desc, X in matrices:
    print(f"   📐 {nombre}: shape {X.shape} — {desc}")
```

## 📐 13. Método del Codo y Coeficiente de Silueta

Una de las preguntas más importantes en K-Means es: **¿cuántos grupos (k) debo usar?**

### Método del Codo (Elbow Method)
Entrena K-Means para múltiples valores de `k` y grafica la **inercia** (suma de distancias cuadráticas de cada punto a su centroide). El "codo" es el punto donde la reducción deja de ser significativa. Es el número óptimo de clústeres.

### Coeficiente de Silueta (Silhouette Score)
Mide qué tan bien separados están los clústeres. Varía entre -1 y 1:
- **Cercano a 1**: Los puntos están bien agrupados en su clúster
- **Cercano a 0**: Los puntos están en el borde entre clústeres
- **Negativo**: Los puntos están mal asignados

Se elige el `k` que **maximiza** el coeficiente de silueta.

```python
# ============================================================
# SECCIÓN 13: Método del Codo y Silueta
# ============================================================

def analizar_k_optimo(X, titulo, k_rango=range(2, 11)):
    inercias = []
    siluetas = []
    k_vals = list(k_rango)

    for k in k_vals:
        km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42, max_iter=300)
        km.fit(X)
        inercias.append(km.inertia_)
        siluetas.append(silhouette_score(X, km.labels_))

    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f'Selección de k óptimo — {titulo}', fontsize=13, fontweight='bold')

    # Método del Codo
    mejor_k = k_vals[np.argmax(siluetas)]
    axes[0].plot(k_vals, inercias, 'o-', color='#6c63ff', linewidth=2.5, markersize=8)
    axes[0].fill_between(k_vals, inercias, alpha=0.1, color='#6c63ff')
    axes[0].axvline(mejor_k, color='#e63946', linestyle='--', alpha=0.7, label=f'k sugerido = {mejor_k}')
    axes[0].set_title('Método del Codo (Elbow Method)')
    axes[0].set_xlabel('Número de clústeres (k)')
    axes[0].set_ylabel('Inercia (WCSS)')
    axes[0].set_xticks(k_vals)
    axes[0].legend()
    axes[0].spines[['top','right']].set_visible(False)

    # Coeficiente de Silueta
    colores_bar = ['#e63946' if k == mejor_k else '#48cae4' for k in k_vals]
    axes[1].bar(k_vals, siluetas, color=colores_bar, edgecolor='white', linewidth=0.8)
    axes[1].set_title('Coeficiente de Silueta (Silhouette Score)')
    axes[1].set_xlabel('Número de clústeres (k)')
    axes[1].set_ylabel('Silhouette Score (mayor = mejor)')
    axes[1].set_xticks(k_vals)
    axes[1].axhline(max(siluetas), color='#e63946', linestyle='--', alpha=0.7,
                    label=f'Máx: {max(siluetas):.3f} en k={mejor_k}')
    axes[1].legend()
    axes[1].spines[['top','right']].set_visible(False)

    plt.tight_layout()
    plt.show()

    print(f"   📊 k óptimo sugerido: {mejor_k} (Silhouette Score = {max(siluetas):.4f})")
    print()
    return mejor_k, inercias, siluetas

print("🔍 Analizando número óptimo de clústeres...\n")
```

## 🤖 14. Entrenamiento y Visualización de Modelos K-Means

Con el número óptimo de clústeres determinado, ahora entrenamos los modelos finales de K-Means.

**Parámetros del modelo K-Means:**

| Parámetro | Valor | Descripción |
|---|---|---|
| `n_clusters` | Variable | Número de grupos a formar |
| `init` | `k-means++` | Inicialización inteligente para evitar mínimos locales |
| `n_init` | `10` | Repeticiones con distintas semillas; se elige la mejor |
| `max_iter` | `300` | Máximo de iteraciones por ejecución |
| `random_state` | `42` | Para reproducibilidad |

```python
# ============================================================
# SECCIÓN 14: Entrenamiento y Visualización de K-Means
# ============================================================

COLORS_CLUSTERS = ['#e63946', '#6c63ff', '#f77f00', '#2dc653', '#48cae4', '#ff9f1c', '#2b2d42']

def entrenar_y_visualizar(X, k, xlabel, ylabel, titulo, datos_tipo='Original'):
    km = KMeans(n_clusters=k, init='k-means++', n_init=10,
                max_iter=300, random_state=42)
    labels = km.fit_predict(X)
    centroids = km.cluster_centers_
    inertia = km.inertia_
    n_iter = km.n_iter_
    sil = silhouette_score(X, labels)

    fig, ax = plt.subplots(figsize=(10, 7))
    patches = []
    for cid in range(k):
        mask = labels == cid
        ax.scatter(X[mask, 0], X[mask, 1],
                   c=COLORS_CLUSTERS[cid], alpha=0.7, s=60,
                   edgecolors='white', linewidth=0.5)
        ax.scatter(centroids[cid, 0], centroids[cid, 1],
                   c=COLORS_CLUSTERS[cid], marker='*', s=350,
                   edgecolors='white', linewidth=1.5, zorder=5)
        ax.annotate(f'C{cid+1}', xy=(centroids[cid, 0], centroids[cid, 1]),
                    xytext=(5, 5), textcoords='offset points',
                    fontweight='bold', fontsize=9)
        patches.append(mpatches.Patch(color=COLORS_CLUSTERS[cid],
                                      label=f'Clúster {cid+1} (n={mask.sum()})'))

    ax.set_title(f'{titulo}\n({datos_tipo} | k={k})', fontsize=13, fontweight='bold')
    ax.set_xlabel(xlabel, fontsize=11)
    ax.set_ylabel(ylabel, fontsize=11)
    ax.legend(handles=patches, loc='upper right', fontsize=9)
    ax.spines[['top','right']].set_visible(False)
    plt.tight_layout()
    plt.show()

    print(f"\n📋 Reporte del Modelo — {titulo} ({datos_tipo}):")
    print(f"   • n_clusters  : {k}")
    print(f"   • Inercia     : {inertia:.2f}")
    print(f"   • Iteraciones : {n_iter}")
    print(f"   • Silhouette  : {sil:.4f}")
    print(f"   • init        : k-means++ | n_init: 10 | max_iter: 300 | random_state: 42")
    print(f"   • Distribución de clústeres:")
    for i in range(k):
        n = (labels == i).sum()
        print(f"     - Clúster {i+1}: {n} clientes ({n/len(labels)*100:.1f}%)")
    print()
    return km, labels

print("✅ Función de modelado definida correctamente.")
```
